```python
import sys
sys.path.append('../src/')
```

```python
from datachecker.engine import ValidationEngine
from datachecker.schema_loader import YamlFileSchemaLoader
from datachecker.plans import PydanticPlanCompiler
from datachecker.batch_sources import CsvBatchSource
from datachecker.batch_validators import PydanticBatchValidator
from datachecker.sinks import JsonlSink  # (you’ll create this)
```

```python

engine = ValidationEngine(
    schema_loader=YamlFileSchemaLoader("../schemas"),
    compiler=PydanticPlanCompiler(),
    source=CsvBatchSource("../data/users.csv", delimiter=";"),
    sink=JsonlSink("../out/report.jsonl"),
    batch_validator=PydanticBatchValidator(),
    batch_size=5,
)
```

```python

engine.run("user")
```
